```python
# Importando a biblioteca do BigQuery

from google.cloud import bigquery
```

```python
#configurando o nome do projeto

client = bigquery.Client(project="portfolio-marketing-digital")

#vai pegar as tabelas, mas não executará nada

table = client.get_table(
    "bigquery-public-data.ga4_obfuscated_sample_ecommerce.events_20210131"
)
```

```python
#irá imprimir o nome da tabela, o número de linhas e o tamanho em MiB
print(f"Linhas: {table.num_rows:,} | Tamanho: {table.num_bytes / 1024**2:.1f} MiB")

print(f"Nome da tabela: {table.table_id}")
```

```python
#Laço para percorrer os campos do esquema da tabela e imprimir o nome, tipo e modo de cada campo

for field in table.schema:
    print(field.name, field.field_type, field.mode)
```
